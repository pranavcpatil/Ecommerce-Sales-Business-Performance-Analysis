```python
# Q1. Calculate total revenue, total cost, and total profit.

total_revenue = order_items["line_revenue"].sum()
total_cost = order_items["line_cost"].sum()
total_profit = order_items["line_profit"].sum()

print("Total Revenue:", round(total_revenue, 2))
print("Total Cost:", round(total_cost, 2))
print("Total Profit:", round(total_profit, 2))
```

```python
# Q2. Which product categories generate the highest revenue?

category_data = order_items.merge(products, on="product_id")

category_revenue = category_data.groupby("category")["line_revenue"].sum()

category_revenue = category_revenue.sort_values(ascending=False)

print(category_revenue)
```

```python
# Q3. Which product categories generate the highest total profit?

category_profit = category_data.groupby("category")["line_profit"].sum()

category_profit = category_profit.sort_values(ascending=False)

print(category_profit)
```

```python
# Q4. What are the top 10 products by revenue?

product_revenue = order_items.groupby("product_id")["line_revenue"].sum()

product_revenue = product_revenue.sort_values(ascending=False).head(10)

top_products = product_revenue.reset_index().merge(products, on="product_id")

print(top_products[["product_name", "line_revenue"]])
```

```python
# Q5. Which sales channel generates the highest revenue?

channel_revenue = orders.merge(order_items, on="order_id")

channel_revenue = channel_revenue.groupby("sales_channel")["line_revenue"].sum()

channel_revenue = channel_revenue.sort_values(ascending=False)

print(channel_revenue)

# Python treats these as different categories, even though they represent the same channel.
# solve--

# orders["sales_channel"] = orders["sales_channel"].str.strip().str.title()
```

```python
# Q6. Which payment methods generate the highest revenue?

payment_revenue = orders.merge(order_items, on="order_id")

payment_revenue = payment_revenue.groupby("payment_method")["line_revenue"].sum()

payment_revenue = payment_revenue.sort_values(ascending=False)

print(payment_revenue)

# This will combine those duplicate categories into consistent names.

orders["payment_method"] = orders["payment_method"].str.strip().str.title()
```

```python
# Q7. How does revenue change month by month? Identify the highest and lowest revenue months.

monthly_revenue = orders.merge(order_items, on="order_id")

monthly_revenue["month"] = monthly_revenue["order_date"].dt.to_period("M")

monthly_revenue = monthly_revenue.groupby("month")["line_revenue"].sum()

print(monthly_revenue)
```

```python
# Now let's identify the highest and lowest revenue months from your result.
print("Highest Revenue Month:")
print(monthly_revenue.idxmax(), monthly_revenue.max())

print("Lowest Revenue Month:")
print(monthly_revenue.idxmin(), monthly_revenue.min())
```

```python
# Q8. Which states generate the highest revenue?

state_revenue = orders.merge(customers, on="customer_id")

state_revenue = state_revenue.merge(order_items, on="order_id")

state_revenue = state_revenue.groupby("state")["line_revenue"].sum()

state_revenue = state_revenue.sort_values(ascending=False)

print(state_revenue)
```

```python
# Q9. Who are the top 10 customers by total revenue?

customer_revenue = orders.merge(order_items, on="order_id")

customer_revenue = customer_revenue.groupby("customer_id")["line_revenue"].sum()

customer_revenue = customer_revenue.sort_values(ascending=False).head(10)

top_customers = customer_revenue.reset_index().merge(customers, on="customer_id")

print(top_customers[["customer_name", "line_revenue"]])
```

```python
# Q10. What is the Average Order Value (AOV) of the business?

total_revenue = order_items["line_revenue"].sum()

total_orders = orders["order_id"].nunique()

aov = total_revenue / total_orders

print("Average Order Value:", round(aov, 2))
```

```python
# Q11. How many orders are there in each order status, and what percentage of total orders does each status represent?

status_count = orders["order_status"].value_counts()

status_percentage = orders["order_status"].value_counts(normalize=True) * 100

status_summary = pd.DataFrame({
    "Order Count": status_count,
    "Percentage": status_percentage
})

print(status_summary.round(2))
```

```python
# Q12. What is the overall order return rate?

total_orders = orders["order_id"].nunique()

returned_orders = (orders["order_status"] == "Returned").sum()

return_rate = (returned_orders / total_orders) * 100

print("Return Rate:", round(return_rate, 2), "%")
```

```python
# Q13. How do different discount levels affect total profit?

# print(order_items.columns)

discount_profit = order_items.groupby("discount_pct")["line_profit"].sum()

print(discount_profit)
    
```

```python
# Q14. What is the yearly revenue, 
yearly_revenue = orders.merge(order_items, on="order_id")

yearly_revenue["year"] = yearly_revenue["order_date"].dt.year

yearly_revenue = yearly_revenue.groupby("year")["line_revenue"].sum()

print(yearly_revenue)


```

```python
# and how much did revenue grow from 2024 to 2025?

revenue_2024 = yearly_revenue.loc[2024]
revenue_2025 = yearly_revenue.loc[2025]

growth = ((revenue_2025 - revenue_2024) / revenue_2024) * 100

print("Revenue Growth:", round(growth, 2), "%")
```

```python
# Q15. Which are the top 10 most profitable products?

product_profit = order_items.groupby("product_id")["line_profit"].sum()

product_profit = product_profit.sort_values(ascending=False).head(10)

top_profitable_products = product_profit.reset_index().merge(products, on="product_id")

print(top_profitable_products[["product_name", "line_profit"]])
```
